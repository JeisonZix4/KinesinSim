## 1. Definir un perfil energético ilustrativo

Los valores siguientes solo construyen un ejemplo gráfico en unidades de kBT. No son parámetros medidos de kinesina.

```python
import matplotlib.pyplot as plt
from protein_motor_sim.visualization.energy_landscape import plot_state_energy_diagram

state_energies = {
    "apo_bound": 0.0,
    "ATP_bound": -0.5,
    "post_hydrolysis": -0.2,
    "stepped": -0.8,
}
transition_barriers = {
    ("apo_bound", "ATP_bound"): 2.0,
    ("ATP_bound", "post_hydrolysis"): 2.2,
    ("post_hydrolysis", "stepped"): 1.7,
}

ax = plot_state_energy_diagram(state_energies, transition_barriers)
plt.show()
```

## 2. Por qué la energía no se obtiene de cualquier tasa

Para relacionar tasas hacia delante y hacia atrás con diferencias de energía libre se necesitan hipótesis termodinámicas, por ejemplo balance detallado/local detallado, además de considerar el acoplamiento químico y mecánico. Una red cinética impulsada por hidrólisis de ATP está fuera del equilibrio.

```python
print("Estados:", list(state_energies))
print("Energías relativas (kBT):", list(state_energies.values()))
print("Barreras especificadas:", transition_barriers)
```

# Superficie de energía en 3D

Este notebook muestra la misma función potencial como una superficie tridimensional. La tercera coordenada representa \(U/(k_B T)\), no una coordenada espacial de la proteína.

```python
from pathlib import Path
import sys

# Find the project root whether JupyterLab was launched from the repository
# root or from the notebooks/ folder.
candidates = [Path.cwd(), *Path.cwd().parents]
PROJECT_ROOT = next(
    (path for path in candidates if (path / "src" / "protein_motor_sim").is_dir()),
    None,
)
if PROJECT_ROOT is not None:
    sys.path.insert(0, str(PROJECT_ROOT / "src"))

print("Project root:", PROJECT_ROOT or "package expected to be installed in this kernel")
```

## 1. Crear la superficie

```python
import matplotlib.pyplot as plt
from protein_motor_sim.mechanics.energy_landscape import EnergyLandscape
from protein_motor_sim.visualization.energy_maps import plot_energy_landscape_3d

landscape = EnergyLandscape(
    period_nm=8.2,
    spatial_barrier_kBT=2.8,
    conformational_barrier_kBT=3.5,
    conformational_bias_kBT=0.3,
    coupling_kBT=0.25,
    opposing_force_pN=0.8,
)
ax, surface = plot_energy_landscape_3d(
    landscape,
    x_range_nm=(-16.4, 24.6),
    resolution=(190, 100),
)
ax.view_init(elev=30, azim=-58)
plt.show()
```

## 2. Cambiar el punto de vista

La superficie puede inspeccionarse desde diferentes ángulos para distinguir los valles conformacionales y las barreras periódicas.

```python
fig = plt.figure(figsize=(10, 7))
ax = fig.add_subplot(111, projection="3d")
plot_energy_landscape_3d(landscape, (-16.4, 24.6), resolution=(150, 80), ax=ax)
ax.view_init(elev=18, azim=35)
plt.show()
```

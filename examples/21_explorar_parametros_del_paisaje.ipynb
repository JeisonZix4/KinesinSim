# Explorar parámetros del paisaje de energía

Este notebook compara cómo las barreras espaciales, las barreras conformacionales y el sesgo energético modifican la geometría del potencial fenomenológico.

```python
from pathlib import Path
import sys

# Find the project root whether JupyterLab was launched from the repository
# root or from the notebooks/ folder.
candidates = [Path.cwd(), *Path.cwd().parents]
PROJECT_ROOT = next(
    (path for path in candidates if (path / "src" / "protein_motor_sim").is_dir()),
    None,
)
if PROJECT_ROOT is not None:
    sys.path.insert(0, str(PROJECT_ROOT / "src"))

print("Project root:", PROJECT_ROOT or "package expected to be installed in this kernel")
```

## 1. Variar la barrera espacial

```python
import matplotlib.pyplot as plt
from protein_motor_sim.mechanics.energy_landscape import EnergyLandscape
from protein_motor_sim.visualization.energy_maps import plot_energy_landscape_2d

barriers = [1.0, 2.5, 5.0]
fig, axes = plt.subplots(1, 3, figsize=(16, 4.8), constrained_layout=True)
for barrier, ax in zip(barriers, axes):
    model = EnergyLandscape(spatial_barrier_kBT=barrier)
    plot_energy_landscape_2d(model, (-8.2, 24.6), resolution=(160, 90), ax=ax)
    ax.set_title(f"Spatial barrier = {barrier:g} kBT")
plt.show()
```

## 2. Variar el sesgo entre conformaciones

```python
fig, ax = plt.subplots(figsize=(10, 5), constrained_layout=True)
for bias in (-1.0, 0.0, 1.0, 2.0):
    model = EnergyLandscape(conformational_bias_kBT=bias)
    profile = model.energy_kBT(0.0, __import__('numpy').linspace(0.0, 1.0, 300))
    ax.plot(__import__('numpy').linspace(0.0, 1.0, 300), profile, label=f"bias={bias:g} kBT")
ax.set_xlabel("Conformational coordinate q")
ax.set_ylabel(r"U / (k$_B$T), at x = 0 nm")
ax.legend()
ax.grid(alpha=0.25)
plt.show()
```

## Interpretación

Los parámetros controlan la forma matemática del mapa. No se deben elegir para forzar una velocidad objetivo: para un modelo predictivo hay que estimarlos o ajustarlos contra datos independientes y comprobar la consistencia termodinámica de las tasas.

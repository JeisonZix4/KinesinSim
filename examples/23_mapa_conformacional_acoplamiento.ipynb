Este notebook explora una segunda formulación fenomenológica, con la coordenada conformacional \(q\) entre -1 (no acoplada) y +1 (acoplada). El mapa ilustra cómo puede modificarse la barrera axial al cambiar la conformación.

## 1. Evaluar la superficie \(U(x,q)\)

```python
import matplotlib.pyplot as plt
import numpy as np
from protein_motor_sim.visualization.energy_landscape import (
    conformational_energy_surface_kbt,
    plot_conformational_energy_surface,
)

ax, contour = plot_conformational_energy_surface(
    x_range_nm=(-16.4, 32.8),
    lattice_spacing_nm=8.2,
    barrier_kbt=5.0,
    conformational_barrier_kbt=2.0,
    docking_bias_kbt=0.8,
    docking_barrier_coupling=0.45,
    opposing_force_pN=0.0,
)
plt.show()
```

## 2. Comparar con distintos niveles de carga

```python
fig, axes = plt.subplots(1, 3, figsize=(16, 4.8), constrained_layout=True)
for force, ax in zip((0.0, 1.5, 3.0), axes):
    plot_conformational_energy_surface(
        x_range_nm=(-8.2, 24.6),
        opposing_force_pN=force,
        x_points=180,
        q_points=100,
        ax=ax,
    )
    ax.set_title(f"Opposing load = {force:g} pN")
plt.show()
```

## Interpretación

La reducción de barrera al aumentar q es un término fenomenológico parametrizado para explorar acoplamiento. No implica que el ciclo real de kinesina siga exactamente esta superficie; para una interpretación cuantitativa se necesitarían coordenadas estructurales y datos de energía libre.

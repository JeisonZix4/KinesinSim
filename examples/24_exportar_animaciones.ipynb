# Exportar animaciones a GIF o MP4

Además de mostrarlas dentro de JupyterLab, podemos guardar las animaciones para informes y presentaciones. GIF utiliza Pillow; MP4 requiere FFmpeg instalado en el sistema.

```python
from pathlib import Path
import sys

# Find the project root whether JupyterLab was launched from the repository
# root or from the notebooks/ folder.
candidates = [Path.cwd(), *Path.cwd().parents]
PROJECT_ROOT = next(
    (path for path in candidates if (path / "src" / "protein_motor_sim").is_dir()),
    None,
)
if PROJECT_ROOT is not None:
    sys.path.insert(0, str(PROJECT_ROOT / "src"))

print("Project root:", PROJECT_ROOT or "package expected to be installed in this kernel")
```

## 1. Generar una animación

```python
from protein_motor_sim import Environment, GillespieSimulator, Kinesin1
from protein_motor_sim.visualization.animation import AnimationStyle, animate_kinesin_walk

trajectory = GillespieSimulator().run(
    Kinesin1(), Environment(), t_max=0.15, seed=44
)
animation = animate_kinesin_walk(
    trajectory,
    style=AnimationStyle(frames_per_step=8, interval_ms=70),
)
```

## 2. Guardar como GIF

El archivo se guarda dentro de `results/animations/`.

```python
from pathlib import Path
from protein_motor_sim.visualization.animation import save_animation

output_dir = Path("results/animations")
output_dir.mkdir(parents=True, exist_ok=True)
gif_path = save_animation(animation, output_dir / "kinesin_walk.gif", fps=18, dpi=110)
print("Saved:", gif_path)
```

## 1. Definir el paisaje energético

El objeto `EnergyLandscape` permite controlar la barrera espacial, la barrera conformacional, el sesgo entre conformaciones, el acoplamiento y una carga opuesta. Sus parámetros son fenomenológicos e ilustrativos.

```python
from protein_motor_sim.mechanics.energy_landscape import EnergyLandscape

landscape = EnergyLandscape(
    period_nm=8.2,
    spatial_barrier_kBT=2.5,
    conformational_barrier_kBT=3.0,
    conformational_bias_kBT=0.0,
    phase_shift=0.5,
    coupling_kBT=0.4,
    opposing_force_pN=0.0,
    temperature_K=310.0,
)
print("kBT =", landscape.thermal_energy_pN_nm, "pN nm")
```

## 2. Mapa de contorno

Los mínimos y barreras son los de la función potencial elegida; no deben interpretarse como un paisaje de energía libre medido para kinesina.

```python
import matplotlib.pyplot as plt
from protein_motor_sim.visualization.energy_maps import plot_energy_landscape_2d

ax, contour = plot_energy_landscape_2d(
    landscape,
    x_range_nm=(-16.4, 32.8),
    q_range=(0.0, 1.0),
    resolution=(280, 150),
    levels=32,
)
plt.show()
```

## 3. Comparar ausencia de carga y carga opuesta

En esta convención, una carga positiva se opone al avance hacia \(+x\), por lo que inclina el potencial hacia energías mayores a posiciones positivas.

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5), constrained_layout=True)
plot_energy_landscape_2d(landscape, (-8.2, 24.6), resolution=(220, 120), opposing_force_pN=0.0, ax=axes[0])
plot_energy_landscape_2d(landscape, (-8.2, 24.6), resolution=(220, 120), opposing_force_pN=2.0, ax=axes[1])
axes[0].set_title("Sin carga externa")
axes[1].set_title("Carga opuesta de 2 pN")
plt.show()
```
